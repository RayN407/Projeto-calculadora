### Calculadora

```python
# Criando mensagens de entrada para os usuários
nome_usuario = input('Por favor, digite seu nome: ')
print(f'Olá {nome_usuario}, Seja bem vindo ao meu projeto de calculadora!')


resposta_apresentação = input('Gostaria de uma pequena apresentação sobre o projeto? ')
while True:
    if resposta_apresentação.lower() == 'sim':
        print('Minha calculadora realiza operações com sete operadores: adição (+), subtração (-), multiplicação (*), divisão (/), divisão inteira (//), resto da divisão (%) e exponenciação (**).')
        print('Para obter o resultado de uma operação, você deve inserir primeiro o valor inicial (valor1), em seguida o segundo valor (valor2) e, por fim, informar um dos operadores listados acima para executar o cálculo desejado.')

        resposta_apresentação2 = input('Entendeu como ela funciona? ')
        if resposta_apresentação2.lower() == 'sim':
            print('Então vamos direto para os cálculos!')
            break
        elif resposta_apresentação2.lower() == 'não' or resposta_apresentação2.lower() == 'nao':
            print('Não se preocupe, é apenas uma calculadora. Vamos aprender na prática!')
            break
        else:
            print('Entrada inválida. Por favor, digite "Sim" ou "Não".')


    elif resposta_apresentação.lower() == 'não' or resposta_apresentação.lower() == 'nao':
        print('Então vamos direto para os cálculos!')
        break
    else:
        resposta_apresentação = input('Entrada inválida. Por favor, digite "Sim" ou "Não": ')

# Loop para multiplas operaçoes
while True:
    # criando as variáveis com tratamento de erro para entrada
    while True:
        try:
            valor1 = float(input('Digite o primeiro valor: '))
            break # Sai do loop se a entrada for válida
        except ValueError:
            print('Entrada inválida. Por favor, digite um número válido.')

    while True:
        try:
            valor2 = float(input('Digite o segundo valor: '))
            break # Sai do loop se a entrada for válida
        except ValueError:
            print('Entrada inválida. Por favor, digite um número válido.')

    operador = input('Digite o operador: ')

    # criando as condiçoes da calculadora
    if operador == '+':              # adição
        resultado = valor1 + valor2

    elif operador == '-':            # subtração
        resultado = valor1 - valor2

    elif operador == '*':            # multiplicação
        resultado = valor1 * valor2

    elif operador == '/':            # divisão
        if valor2 == 0:
            print('Não é possível dividir um número por zero')
        else:
            resultado = valor1 / valor2

    elif operador == '//':           # divisão inteira
        if valor2 == 0:
            print('Não é possível dividir um número por zero')
        else:
            resultado = valor1 // valor2

    elif operador == '%':            # resto da divisão
        if valor2 == 0:
            print('Não é possível calcular o resto da divisão por zero')
        else:
            resultado = valor1 % valor2

    elif operador == '**':           # exponenciação
        resultado = valor1 ** valor2

    else:                            # caso informe um caracter que não está em uma das condições retorna como inválido
        print(f'Operador inválido: {operador}')

    # executando a calculadora e imprimindo um resultado se houver
    if resultado is not None:
        print(f'O resultado da operação é: {resultado}')

    # Pergunta se o usuário quer fazer outra operação
    continuar = input('Deseja realizar outra operação? : ')
    if continuar.lower() == 'sim':
        print('Então vamos para um novo cálculo!')
    elif continuar.lower() == 'não' or continuar.lower() == 'nao':
        print('Obrigado por usar a calculadora!')
        break # Sai do loop principal se o usuário não quiser continuar
    else:
        print('Entrada inválida. Por favor, digite "Sim" ou "Não".')

```
